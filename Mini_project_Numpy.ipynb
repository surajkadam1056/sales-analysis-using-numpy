```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200, 170, 210],
    [200, 220, 190, 250, 230, 270],
    [90, 110, 130, 100, 140, 150],
    [300, 280, 320, 350, 330, 370]
])

products = np.array([
    "Laptop",
    "Mobile",
    "Headphones",
    "Tablet"
])

months = np.array([
    "Jan", "Feb", "Mar", "Apr", "May", "Jun"
])
```

# Task 3
Part 2 — Sales Analysis

Find:

1)Total sales of all products
2)Average sales
3)Maximum sales
4)Minimum sales
5)Total sales product-wise
6)Total sales month-wise
7)Average sales product-wise
8)Average sales month-wise


```python
# Task 3 answers :
#1) 
Total_sales= np.sum(sales)
print(Total_sales)
#2)
avg_sales=np.average(sales)
print(avg_sales)
#3) 
min_sales=np.min(sales)
print(min_sales)
#4)
max_sales=np.max(sales)
print(max_sales)
#5)
total_sales_product_wise = np.sum(sales, axis=1)
print(total_sales_product_wise)
#6)
total_sales_months_wise= np.sum(sales,axis=0)
print(total_sales_months_wise)
#7)
Avg_sales_product_wise = np.average(sales, axis=1)
print(Avg_sales_product_wise)
#8)
Avg_sales_months_wise= np.average(sales,axis=0)
print(Avg_sales_months_wise)
```

 # Task 5 
Part 4 — Business Logic

1)200 se greater sales ko "High" aur baaki ko "Low" label karo.
2)Har product ki best monthly sales find karo.
3)Har product ki worst monthly sales find karo.
4)Overall highest sales value find karo.

```python
# Task 5 answers:
#1)
print(np.where(sales>200,"High","Low"))
#2)
best_monthly_sales=np.max(sales, axis=1)
print(best_monthly_sales)
#3)
worst_monthly_sales= np.min(sales,axis=1)
print(worst_monthly_sales)
#4) 
highest_sales=np.max(sales)
print(highest_sales)
```

# Task 6 :
Part 5 — Broadcasting

Suppose company ne next month expected sales mein 10% growth assume ki.

Then find :

1)Expected sales
2)Additional expected sales
3)Product-wise expected sales

```python
# task 6 answrers:
#1)
expected_sales = sales*1.10
print(expected_sales)

#2) additional_sales = expected_sale - current_sales
additional_sales= sales*1.10 - sales 
print(additional_sales)

#3)
product_wise_expected_sales = np.sum(sales*1.10, axis=1)
print(product_wise_expected_sales)
```

# Task 7 
Business Questions

Q1. Which product has the highest total sales?

Q2. Which month has the highest total sales?

Q3. Overall average sales kitni hai?

Q4. Kitni sales 200 se greater hain?

Q5. 10% growth ke baad total expected sales kitni hogi?


```python
# task 7 answers 
#1)
product_total = np.sum(sales, axis=1)
highest_product = products[np.argmax(product_total)]
print(highest_product)

#2)
month_total = np.sum(sales, axis=0)
highest_month = months[np.argmax(month_total)]
print(highest_month)

#3)
average_sales = np.mean(sales)
print(average_sales)

#4)
count = np.sum(sales > 200)
print(count)

#5)
total_expected_sales = np.sum(expected_sales)
print(total_expected_sales)
```
